## Machine Learning Project on Topic Modeling
Topic modeling can be seen as a task of machine learning which can be used to present the huge volume of data generated due to advancements in computer and web technology in low dimension and to present the hidden concepts, important characteristics or latent variables of the data, depending on the context of the application of the identified text.

```python
from bertopic import BERTopic
from sklearn.datasets import fetch_20newsgroups

docs = fetch_20newsgroups(
    subset='all',
    remove=('headers', 'footers', 'quotes')
)['data']

# boş olanları temizleyelim
docs = [doc for doc in docs if doc.strip() != ""]
```

```python
topic_model = BERTopic(
    language="english",
    calculate_probabilities=True,
    verbose=True
)
```

```python
topics, probs = topic_model.fit_transform(docs)
```

```python
topic_model.get_topic_info().head(10)
```

```python
topic_model.get_topic(0)
```

```python
for i in range(5):
    print("TOPIC:", topics[i])
    print(docs[i][:300])
    print("-" * 80)
```

```python
topic_model.visualize_topics()
```

```python
topic_model.visualize_barchart()
topic_model.visualize_hierarchy()
```

```python
topic_model.save(
    "bertopic_20newsgroups",
    serialization="pickle"
)
```

```python
from bertopic import BERTopic

topic_model = BERTopic.load("bertopic_20newsgroups")

text = "NASA launched a new satellite into space"
topic, prob = topic_model.transform([text])

print("Topic:", topic)
```
