```python
import pickle
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import linear_kernel
import seaborn as sns
```

```python
sns.heatmap(df.corr(numeric_only=True))
```

Sadece gerekli sütunları al ve boş verileri temizle

```python
df = df[['book_title', 'book_desc', 'book_rating_count']]
df = df.dropna()
```

```python
df = df.sort_values('book_rating_count', ascending=False).head(5000)
df = df.reset_index(drop=True)
```

TF-IDF ile Metin Vektörleştirme

```python
tfidf = TfidfVectorizer(stop_words='english')
tfidf_matrix = tfidf.fit_transform(df['book_desc'])
```

```python
similarity = linear_kernel(tfidf_matrix, tfidf_matrix)
```

```python
pickle.dump(df, open('books_list.pkl', 'wb'))
```

```python
pickle.dump(similarity, open('similarity.pkl', 'wb'))
```
